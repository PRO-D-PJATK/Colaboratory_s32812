## 6) Simple linear regression
Goal: predict **price (thousands of PLN)** from **floor area (m²)**.
Steps:
1. Split features X (area) and target y (price).
2. Fit `LinearRegression`.
3. Inspect parameters (coefficient and intercept).
4. Plot the fitted line.


````python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# X: floor area (m²), y: price (thousands of PLN)
X = df[["area_m2"]].values
y = df["price_thousands_pln"].values

model = LinearRegression()
model.fit(X, y)

print("Coefficient (a):", model.coef_[0])
print("Intercept (b):", model.intercept_)

# Predictions for a few sample areas
sample = np.array([[40],[60],[80],[100]])
pred = model.predict(sample)
for m2, p in zip(sample.flatten(), pred):
    print(f"Predicted price for {m2} m²: {p:.1f} thousand PLN")

# Plot: points + regression line
plt.figure()
plt.scatter(X, y, label="Data")
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1,1)
y_line = model.predict(x_line)
plt.plot(x_line, y_line, label="Linear regression")
plt.xlabel("Floor area (m²)")
plt.ylabel("Price (thousands of PLN)")
plt.title("Simple linear regression: price = a*area + b")
plt.legend()
plt.show()
````

### Your exercises
1. Add a new column with **price per m²** (`price_per_m2`) and inspect how it changes across areas.
2. Add mild non-linearity (e.g. `area_m2^2` as an extra feature) and check whether the error decreases.
3. Save the plot as PNG (e.g. `plt.savefig("regression.png")`).

> Extension: split into **train/test** (`train_test_split`) and compute error metrics (e.g. MAE, RMSE).


````python
# Exercise 1: price per m2 column
df["price_per_m2"] = df["price_thousands_pln"] / df["area_m2"]
display(df)
print(df["price_per_m2"].describe())
print("Correlation area vs price_per_m2:", df["area_m2"].corr(df["price_per_m2"]))
````

````python
# Exercise 2: area vs price per m2
plt.figure()
plt.plot(df["area_m2"], df["price_per_m2"], marker="o")
plt.xlabel("Floor area (m²)")
plt.ylabel("Price per m² (thousands of PLN)")
plt.title("Price per m² vs floor area")
plt.grid(True)
plt.show()
````

````python
# Exercise 3: model errors (linear model, whole dataset)
from sklearn.metrics import mean_absolute_error, mean_squared_error  # metrics
y_pred = model.predict(X)
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
print(f"MAE:  {mae:.3f} thousand PLN")
print(f"MSE:  {mse:.3f}")
print(f"RMSE: {rmse:.3f} thousand PLN")
````

````python
# Exercise 4: add area_m2^2 as a second feature (polynomial model)
df["area_m2_sq"] = df["area_m2"] ** 2
X_poly = df[["area_m2", "area_m2_sq"]].values
model_poly = LinearRegression()
model_poly.fit(X_poly, y)
y_pred_poly = model_poly.predict(X_poly)
mae_poly = mean_absolute_error(y, y_pred_poly)
mse_poly = mean_squared_error(y, y_pred_poly)
print("Coefficients (a1, a2):", model_poly.coef_)
print("Intercept (b):", model_poly.intercept_)
print(f"Linear     -> MAE: {mae:.3f}, MSE: {mse:.3f}")
print(f"Polynomial -> MAE: {mae_poly:.3f}, MSE: {mse_poly:.3f}")
````

````python
# Exercise 5: save the regression plot as PNG (linear vs polynomial)
plt.figure()
plt.scatter(X, y, label="Data")
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
plt.plot(x_line, model.predict(x_line), label="Linear regression")
x_line_poly = np.hstack([x_line, x_line ** 2])
plt.plot(x_line, model_poly.predict(x_line_poly), label="Polynomial (area + area^2)")
plt.xlabel("Floor area (m²)")
plt.ylabel("Price (thousands of PLN)")
plt.title("Linear vs polynomial regression")
plt.legend()
plt.savefig("regression.png", dpi=150, bbox_inches="tight")
plt.show()
!ls -l regression.png
````

````python
# Extension: train/test split and error metrics
from sklearn.model_selection import train_test_split  # split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model_tt = LinearRegression()
model_tt.fit(X_train, y_train)
y_test_pred = model_tt.predict(X_test)
mae_test = mean_absolute_error(y_test, y_test_pred)
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
print(f"Train size: {len(X_train)}, test size: {len(X_test)}")
print(f"Test MAE:  {mae_test:.3f} thousand PLN")
print(f"Test RMSE: {rmse_test:.3f} thousand PLN")
````
